```python
import os
import glob
import rasterio
from rasterio.merge import merge

# Folder containing the 4 tiles
tile_folder = r"C:\Users\snvra\NISAR_Data_project\Data\Pich_SatEmb_2025"

# Output mosaic
output_mosaic = r"C:\Users\snvra\NISAR_Data_project\Data\Pich_SatEmb_2025\Pichavaram_Mosaic.tif"

# Read all tif files
tile_files = sorted(glob.glob(os.path.join(tile_folder, "*.tif")))

print("Tiles found:")
for f in tile_files:
    print(os.path.basename(f))

# Open all tiles
src_files = [rasterio.open(f) for f in tile_files]

# Create mosaic
mosaic, out_transform = merge(src_files)

# Copy metadata from first tile
out_meta = src_files[0].meta.copy()

out_meta.update({
    "driver": "GTiff",
    "height": mosaic.shape[1],
    "width": mosaic.shape[2],
    "transform": out_transform
})

# Save mosaic
with rasterio.open(output_mosaic, "w", **out_meta) as dest:
    dest.write(mosaic)

# Close all files
for src in src_files:
    src.close()

print("\nMosaic created successfully!")
print("Saved to:", output_mosaic)
print("Bands:", mosaic.shape[0])
print("Rows:", mosaic.shape[1])
print("Cols:", mosaic.shape[2])
```
